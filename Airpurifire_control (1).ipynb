## Addressing Question 1: Top 5 and bottom 5 areas with highest average AQI

**Question 1**: Top 5 and bottom 5 areas with highest average AQI (considering areas with data covering Dec 2024 → May 2025, i.e., last 6 months).

To answer this, I will perform the following steps:
1.  Convert the 'Date' column in `aqi_df` to datetime objects.
2.  Filter the data to include only the period from December 2024 to May 2025.
3.  Calculate the average AQI for each 'Area' within this filtered period.
4.  Identify the top 5 and bottom 5 areas based on their average AQI.

## Addressing Question 2: Top 2 and bottom 2 prominent pollutants for each state of Southern India

**Question 2**: Top 2 and bottom 2 prominent pollutants for each state of Southern India (states = Andhra Pradesh, Telangana, Karnataka, Kerala, Tamil Nadu; consider 2022 onward).

To answer this, I will perform the following steps:
1.  Define the list of Southern Indian states.
2.  Filter the `aqi_df` for these states and for dates from 2022 onwards.
3.  Identify prominent pollutants from the available columns.
4.  For each state, calculate the average concentration of each pollutant.
5.  Determine the top 2 and bottom 2 pollutants based on their average concentrations for each state.

```python
# 1. Define Southern Indian states
southern_states = [
    'Andhra Pradesh',
    'Telangana',
    'Karnataka',
    'Kerala',
    'Tamil Nadu'
]

# 2. Filter data for southern states and from 2022 onwards
start_date_q2 = pd.to_datetime('2022-01-01')
filtered_aqi_q2 = aqi_df[
    (aqi_df['state'].isin(southern_states)) &
    (aqi_df['date'] >= start_date_q2)
].copy()

# 3. Analyze prominent pollutants by counting their occurrences
# Group by state and count the occurrences of each prominent_pollutants
pollutant_counts_by_state = filtered_aqi_q2.groupby(['state', 'prominent_pollutants']).size().unstack(fill_value=0)

print("Count of Prominent Pollutants by Southern Indian State (from 2022):")
display(pollutant_counts_by_state)

# 4. Determine top 2 and bottom 2 prominent pollutants for each state
prominent_pollutants_results = {}

for state in southern_states:
    if state in pollutant_counts_by_state.index:
        state_pollutants = pollutant_counts_by_state.loc[state]
        # Sort pollutants by count
        sorted_pollutants = state_pollutants.sort_values(ascending=False)

        # Get top 2 (if available)
        top_2 = sorted_pollutants.head(2).index.tolist()

        # Get bottom 2 (if available, and ensuring they are not the same as top 2 if there are few pollutants)
        # This logic needs to be robust for cases with few distinct pollutants or all having the same count
        bottom_2_candidates = sorted_pollutants.tail(2).index.tolist()

        # Filter out pollutants that are already in top_2 if there are more than 2 distinct pollutants
        # And ensure we don't pick the same pollutant for both top and bottom if there's only one type
        if len(sorted_pollutants) > 2:
            temp_bottom = [p for p in bottom_2_candidates if p not in top_2]
            # If temp_bottom is empty, it means top_2 and bottom_2_candidates overlap completely or are the same.
            # In this case, we might need to adjust or simply not report bottom 2 if there isn't clear distinction.
            if len(temp_bottom) == 0 and len(sorted_pollutants) > 0:
                # If all pollutants have the same count, then there's no clear 'bottom'
                if len(set(sorted_pollutants.values)) == 1:
                    bottom_2 = [] # No distinct bottom if all are same
                else:
                    # Try to get the next distinct values from bottom if possible
                    unique_values = sorted_pollutants.unique()
                    if len(unique_values) > 1:
                        second_last_val = unique_values[-2]
                        bottom_2 = sorted_pollutants[sorted_pollutants == second_last_val].index.tolist()
                        if len(bottom_2) > 2:
                            bottom_2 = bottom_2[-2:] # Take last two if many have the same second lowest
                        elif len(bottom_2) == 1 and len(unique_values) > 2: # If only one at second lowest, try third lowest
                            third_last_val = unique_values[-3]
                            bottom_2.extend(sorted_pollutants[sorted_pollutants == third_last_val].index.tolist()[:1])
                    else: # All counts are identical
                        bottom_2 = []
            else:
                bottom_2 = temp_bottom
                if len(bottom_2) < 2 and len(sorted_pollutants) > len(top_2):
                    # If after removing top_2, we still need more for bottom_2, add from the remaining unique lowest ones
                    remaining_pollutants = sorted_pollutants[~sorted_pollutants.index.isin(top_2)].tail(2).index.tolist()
                    for p in remaining_pollutants:
                        if p not in bottom_2:
                            bottom_2.append(p)
                    bottom_2 = bottom_2[-2:] # Ensure only up to 2 are kept

        elif len(sorted_pollutants) <= 2: # If 2 or fewer distinct pollutants, top and bottom might be the same or one
            if len(sorted_pollutants) == 1:
                bottom_2 = [] # Only one pollutant, no distinct bottom
            else:
                if sorted_pollutants.iloc[0] == sorted_pollutants.iloc[1]:
                    bottom_2 = [] # Both are equally prominent, no distinct bottom
                else:
                    bottom_2 = sorted_pollutants.tail(1).index.tolist() # Only one bottom pollutant
        else:
            bottom_2 = sorted_pollutants.tail(2).index.tolist()

        prominent_pollutants_results[state] = {
            'Top 2 Prominent Pollutants (by frequency)': top_2,
            'Bottom 2 Prominent Pollutants (by frequency)': bottom_2
        }
    else:
        prominent_pollutants_results[state] = {
            'Top 2 Prominent Pollutants (by frequency)': 'No data',
            'Bottom 2 Prominent Pollutants (by frequency)': 'No data'
        }

print("\nTop 2 and Bottom 2 Prominent Pollutants for each Southern Indian State (by frequency of occurrence):")
for state, pollutants in prominent_pollutants_results.items():
    print(f"\nState: {state}")
    print(f"  Top 2: {', '.join(pollutants['Top 2 Prominent Pollutants (by frequency)'] )}")
    print(f"  Bottom 2: {', '.join(pollutants['Bottom 2 Prominent Pollutants (by frequency)'] )}")
```

## Addressing Question 3: AQI values on weekends vs weekdays in Indian metro cities

**Question 3**: Do AQI values improve on weekends vs weekdays in Indian metro cities: Delhi, Mumbai, Chennai, Kolkata, Bengaluru, Hyderabad, Ahmedabad, Pune (use data from last 1 year from run date — code uses 365 days prior to the max date in the AQI file).

To answer this, I will perform the following steps:
1.  Define the list of target metro cities.
2.  Determine the latest date in the `aqi_df` to establish the 'run date'.
3.  Filter the `aqi_df` for the last 365 days leading up to the max date and for the specified metro cities.
4.  Extract the day of the week for each entry and categorize them as 'Weekday' or 'Weekend'.
5.  Calculate the average AQI for weekdays and weekends for each city.
6.  Compare and present the results to see if AQI values generally improve (i.e., decrease) on weekends.

```python
import numpy as np

# 1. Define target metro cities
metro_cities = [
    'Delhi',
    'Mumbai',
    'Chennai',
    'Kolkata',
    'Bengaluru',
    'Hyderabad',
    'Ahmedabad',
    'Pune'
]

# Ensure 'area' column is used for filtering city names
# Make city names consistent if needed, assuming 'area' contains city names

# 2. Determine the latest date in the aqi_df
max_date_aqi = aqi_df['date'].max()

# 3. Filter data for the last 365 days and for metro cities
start_date_q3 = max_date_aqi - pd.Timedelta(days=365)
filtered_aqi_q3 = aqi_df[
    (aqi_df['date'] >= start_date_q3) &
    (aqi_df['date'] <= max_date_aqi) &
    (aqi_df['area'].str.contains('|'.join(metro_cities), case=False, na=False)) # Case-insensitive search for cities within area names
].copy()

# Map common city names to standardized metro_cities list for consistency
def standardize_city_name(area_name, metro_cities_list):
    for city in metro_cities_list:
        if city.lower() in area_name.lower():
            return city
    return area_name # Return original if no match

filtered_aqi_q3['city'] = filtered_aqi_q3['area'].apply(lambda x: standardize_city_name(x, metro_cities))
filtered_aqi_q3 = filtered_aqi_q3[filtered_aqi_q3['city'].isin(metro_cities)] # Filter to only exact metro cities

# 4. Extract day of the week and categorize as 'Weekday' or 'Weekend'
# Monday=0, Sunday=6
filtered_aqi_q3['day_of_week'] = filtered_aqi_q3['date'].dt.dayofweek
filtered_aqi_q3['day_type'] = np.where(
    filtered_aqi_q3['day_of_week'].isin([5, 6]), # 5 is Saturday, 6 is Sunday
    'Weekend',
    'Weekday'
)

# 5. Calculate average AQI for weekdays and weekends for each city
aqi_weekday_weekend = filtered_aqi_q3.groupby(['city', 'day_type'])['aqi_value'].mean().unstack()

print("Average AQI on Weekdays vs. Weekends for Indian Metro Cities (Last 1 Year):")
display(aqi_weekday_weekend)

# 6. Compare AQI values and determine if they improve on weekends
aqi_weekday_weekend['Improvement_on_Weekend'] = aqi_weekday_weekend['Weekday'] > aqi_weekday_weekend['Weekend']

print("\nAQI Improvement on Weekends (lower AQI on Weekend is an improvement):")
display(aqi_weekday_weekend[['Weekday', 'Weekend', 'Improvement_on_Weekend']])
```

## Addressing Question 6: Severity Mapping - Persistent or Worsening AQI

**Question 6**: Severity Mapping: identify cities with persistent or worsening AQI (trend analysis & count of days in "Unhealthy" categories).

To answer this, I will perform the following steps:
1.  Define 'Unhealthy' AQI categories based on `air_quality_status` (e.g., 'Poor', 'Very Poor', 'Severe').
2.  Calculate the number of days each city spent in these 'Unhealthy' categories.
3.  Perform a simple trend analysis by comparing the number of 'Unhealthy' days in two consecutive periods (e.g., last 6 months vs. prior 6 months, or year-over-year comparison).
4.  Identify cities that show an increasing trend in 'Unhealthy' days or consistently high counts.

For simplicity and to demonstrate a trend, I'll compare the count of 'Unhealthy' days in the last 6 months of data with the preceding 6 months. I'll also consider cities with a high total count of 'Unhealthy' days.

```python
# 1. Define 'Unhealthy' AQI categories
unhealthy_categories = ['Poor', 'Very Poor', 'Severe', 'Critical', 'Hazardous'] # Add categories as per AQI standards if available

# Filter AQI data to only include 'Unhealthy' days
unhealthy_aqi_df = aqi_df[aqi_df['air_quality_status'].isin(unhealthy_categories)].copy()

# 2. Get the latest date in the dataset for a relevant period comparison
max_date = aqi_df['date'].max()

# Define comparison periods: Last 6 months vs. Prior 6 months
end_period_1 = max_date
start_period_1 = max_date - pd.Timedelta(days=180) # Last 6 months

end_period_2 = start_period_1 - pd.Timedelta(days=1) # Day before start_period_1
start_period_2 = end_period_2 - pd.Timedelta(days=180) # Prior 6 months

# Filter unhealthy data for Period 1 (Last 6 months)
unhealthy_period_1 = unhealthy_aqi_df[
    (unhealthy_aqi_df['date'] >= start_period_1) &
    (unhealthy_aqi_df['date'] <= end_period_1)
]

# Filter unhealthy data for Period 2 (Prior 6 months)
unhealthy_period_2 = unhealthy_aqi_df[
    (unhealthy_aqi_df['date'] >= start_period_2) &
    (unhealthy_aqi_df['date'] <= end_period_2)
]

# Count unhealthy days per city for each period
counts_period_1 = unhealthy_period_1.groupby('area').size().rename('Unhealthy_Days_Last_6_Months')
counts_period_2 = unhealthy_period_2.groupby('area').size().rename('Unhealthy_Days_Prior_6_Months')

# Combine counts for comparison
severity_mapping = pd.concat([counts_period_1, counts_period_2], axis=1).fillna(0)

# Calculate the change in unhealthy days
severity_mapping['Change_Unhealthy_Days'] = severity_mapping['Unhealthy_Days_Last_6_Months'] - severity_mapping['Unhealthy_Days_Prior_6_Months']

# Identify cities with worsening AQI (increased unhealthy days)
worsening_cities = severity_mapping[severity_mapping['Change_Unhealthy_Days'] > 0].sort_values(by='Change_Unhealthy_Days', ascending=False)

# Identify cities with persistently high unhealthy AQI (e.g., top 10 by total unhealthy days in last period)
persistent_unhealthy_cities = severity_mapping.sort_values(by='Unhealthy_Days_Last_6_Months', ascending=False).head(10)

print(f"Cities with a worsening trend in Unhealthy AQI (comparing last 6 months to prior 6 months, periods ending {max_date.strftime('%Y-%m-%d')} and {end_period_2.strftime('%Y-%m-%d')} respectively):\n")
display(worsening_cities)

print(f"\nCities with persistently high Unhealthy AQI (Top 10 by count of unhealthy days in the last 6 months ending {max_date.strftime('%Y-%m-%d')}):\n")
display(persistent_unhealthy_cities)
```

## Addressing Question 7: Health Impact Correlation

**Question 7**: Correlate AQI spikes with health event counts from idsp (e.g., respiratory/ARI/diarrheal?) — produce aggregated correlation measures per state/month.

To answer this, I will perform the following steps:
1.  Process `idsp_df`:
    *   Convert date columns to datetime objects.
    *   Extract 'year' and 'month'.
    *   Filter for specified health events (Acute Diarrheal Disease, Acute Respiratory Infection, Pneumonia, etc.).
    *   Aggregate `cases` per state, year, and month.
2.  Process `aqi_df`:
    *   Ensure 'date' column is datetime.
    *   Extract 'year' and 'month'.
    *   Aggregate `aqi_value` (e.g., mean) per state, year, and month.
3.  Merge Data: Combine the aggregated `aqi_df` and `idsp_df` on 'state', 'year', and 'month'.
4.  Calculate Correlation: Compute the correlation between AQI and health event counts for each state.

```python
# 1. Process idsp_df
# Convert date column to datetime objects (handling potential multiple date formats)
# Using 'reporting_date' as it's more direct for health events
idsp_df['reporting_date'] = pd.to_datetime(idsp_df['reporting_date'], errors='coerce', dayfirst=True)

# Drop rows where date conversion failed
idsp_df_cleaned = idsp_df.dropna(subset=['reporting_date']).copy()

idsp_df_cleaned['year'] = idsp_df_cleaned['reporting_date'].dt.year
idsp_df_cleaned['month'] = idsp_df_cleaned['reporting_date'].dt.month

# Define target health events (assuming these are the relevant ones from the dataset)
target_diseases = [
    'Acute Diarrheal Disease',
    'Acute Respiratory Infection',
    'Pneumonia',
    'Influenza',
    'Bronchitis',
    'Asthma',
    'Cholera',
    'Typhoid'
]

filtered_idsp_q7 = idsp_df_cleaned[idsp_df_cleaned['disease_illness_name'].isin(target_diseases)]

# Aggregate cases per state, year, and month
aggregated_idsp_q7 = filtered_idsp_q7.groupby(['state', 'year', 'month'])['cases'].sum().reset_index()
aggregated_idsp_q7.rename(columns={'cases': 'total_health_cases'}, inplace=True)

print("Aggregated Health Cases (IDSP data per state/year/month):")
display(aggregated_idsp_q7.head())

# 2. Process aqi_df
# 'date' column is already datetime from previous steps and format='%d-%m-%Y'
aqi_df['year'] = aqi_df['date'].dt.year
aqi_df['month'] = aqi_df['date'].dt.month

# Aggregate mean AQI per state, year, and month
aggregated_aqi_q7 = aqi_df.groupby(['state', 'year', 'month'])['aqi_value'].mean().reset_index()
aggregated_aqi_q7.rename(columns={'aqi_value': 'mean_aqi'}, inplace=True)

print("\nAggregated Mean AQI (per state/year/month):")
display(aggregated_aqi_q7.head())

# 3. Merge Data
merged_data_q7 = pd.merge(
    aggregated_aqi_q7,
    aggregated_idsp_q7,
    on=['state', 'year', 'month'],
    how='inner' # Use inner join to only include periods where both AQI and health data exist
)

print("\nMerged AQI and Health Data (first 5 rows):")
display(merged_data_q7.head())

# 4. Calculate Correlation
# Calculate correlation for each state between mean_aqi and total_health_cases
correlation_by_state = merged_data_q7.groupby('state').apply(
    lambda x: x['mean_aqi'].corr(x['total_health_cases'])
)

correlation_by_state = correlation_by_state.rename('AQI_Health_Correlation').dropna()

print("\nCorrelation between Mean AQI and Total Health Cases per State:")
display(correlation_by_state.sort_values(ascending=False))
```

## Addressing Question 8: Demand Triggers - AQI Spikes and Inferred Demand

**Question 8**: Demand Triggers: analyze temporal relationship between pollution spikes (AQI) and inferred demand triggers (proxy: spikes in AQI correlated with news/health events or vehicle counts growth; we’ll show an approach using vehicle registrations & population as proxy for potential market).

This question is multifaceted. Given the available datasets, I will focus on correlating AQI with vehicle counts (from `vahan_df`) and possibly health events (from `idsp_df`) as proxies for demand triggers, aggregated at a state and monthly level.

To answer this, I will perform the following steps:
1.  Process `vahan_df`:
    *   Clean and convert date columns.
    *   Aggregate `total_vehicles` per state, year, and month.
2.  Re-use `aggregated_aqi_q7` (mean AQI per state, year, month) from Question 7.
3.  Merge `aggregated_aqi_q7` with aggregated `vahan_df` on state, year, and month.
4.  Calculate correlation between AQI and vehicle counts for each state.
5.  Optionally, correlate AQI with health events (`aggregated_idsp_q7`) at a state/month level for additional insights into demand triggers.

```python
import pandas as pd

# 1. Process vahan_df
# Ensure 'year' and 'month' columns are available and correctly formatted

vahan_df_cleaned = vahan_df.copy()

# First, create a temporary date string from 'year' and 'month' (e.g., '2022-April-01')
vahan_df_cleaned['temp_date'] = vahan_df_cleaned['year'].astype(str) + '-' + vahan_df_cleaned['month'] + '-01'
# Convert to datetime objects and then extract month as integer
vahan_df_cleaned['month_num'] = pd.to_datetime(vahan_df_cleaned['temp_date']).dt.month

# Aggregate total_vehicles per state, year, and numerical month
aggregated_vahan_q8 = vahan_df_cleaned.groupby(['state', 'year', 'month_num'])['value'].sum().reset_index()
aggregated_vahan_q8.rename(columns={'value': 'total_vehicle_registrations', 'month_num': 'month'}, inplace=True)

print("Aggregated Vehicle Registrations (Vahan data per state/year/month):")
display(aggregated_vahan_q8.head())

# 2. Re-use aggregated_aqi_q7 from Question 7
# 'aggregated_aqi_q7' is already computed: mean_aqi per state, year, month

# 3. Merge aggregated_aqi_q7 with aggregated_vahan_q8
merged_data_q8_vahan = pd.merge(
    aggregated_aqi_q7,
    aggregated_vahan_q8,
    on=['state', 'year', 'month'],
    how='inner'
)

print("\nMerged AQI and Vehicle Registrations Data (first 5 rows):")
display(merged_data_q8_vahan.head())

# 4. Calculate correlation between AQI and vehicle counts
correlation_aqi_vahan_by_state = merged_data_q8_vahan.groupby('state').apply(
    lambda x: x['mean_aqi'].corr(x['total_vehicle_registrations'])
)

correlation_aqi_vahan_by_state = correlation_aqi_vahan_by_state.rename('AQI_Vehicle_Correlation').dropna()

print("\nCorrelation between Mean AQI and Total Vehicle Registrations per State:")
display(correlation_aqi_vahan_by_state.sort_values(ascending=False))

# 5. Optional: Correlate AQI with health events for additional insights (re-using aggregated_idsp_q7)
merged_data_q8_health = pd.merge(
    aggregated_aqi_q7,
    aggregated_idsp_q7,
    on=['state', 'year', 'month'],
    how='inner'
)

correlation_aqi_health_by_state = merged_data_q8_health.groupby('state').apply(
    lambda x: x['mean_aqi'].corr(x['total_health_cases'])
)

correlation_aqi_health_by_state = correlation_aqi_health_by_state.rename('AQI_Health_Correlation_Q8').dropna()

print("\n(Re-calculated) Correlation between Mean AQI and Total Health Cases per State (as a demand trigger proxy):\n(Note: This is similar to Q7, but included here for completeness of demand triggers analysis.)")
display(correlation_aqi_health_by_state.sort_values(ascending=False))
```

## Addressing Question 9: Market Size Proxies

**Question 9**: Market size proxies: leveraging population and other relevant data (e.g., economic indicators like vehicle registrations, household income, urbanization rates) to estimate potential market size.

For this question, I will primarily use the `population_df` to establish a direct proxy for market size. A larger population in a given state generally indicates a larger potential market. I will process the population data to get total population by state and year.

If more granular data or other economic indicators (like per capita income or urbanization rates) were available, they could be integrated for a more sophisticated market sizing model. For now, we'll focus on population projections as a foundational proxy.

```python
# 1. Process population_df
# Filter for 'Total' gender to get overall population counts
population_total_q9 = population_df[population_df['gender'] == 'Total'].copy()

# Ensure 'value' is numeric (it represents population in thousands)
population_total_q9['value'] = pd.to_numeric(population_total_q9['value'], errors='coerce')

# Drop any rows where value could not be converted
population_total_q9.dropna(subset=['value'], inplace=True)

# Aggregate population per state and year
# Summing `value` which is already 'in Thousands' to get total population in thousands
aggregated_population_q9 = population_total_q9.groupby(['state', 'year'])['value'].sum().reset_index()
aggregated_population_q9.rename(columns={'value': 'total_population_thousands'}, inplace=True)

print("Aggregated Total Population (in thousands) per State and Year:")
display(aggregated_population_q9.head())

# We can also look at the latest projected population to identify the largest markets by population.
latest_year = aggregated_population_q9['year'].max()
population_latest_year = aggregated_population_q9[aggregated_population_q9['year'] == latest_year]

largest_markets_by_population = population_latest_year.sort_values(by='total_population_thousands', ascending=False)

print(f"\nTop 10 Largest Potential Markets by Total Population (in thousands) for the latest available year ({latest_year}):")
display(largest_markets_by_population.head(10))

# Further analysis could involve merging this with AQI data to identify large markets with poor air quality.
```
